## Imputation Techniques-

### 1. Mean Value Imputation

```python
df['Age_mean']= df['age'].fillna(df['age'].mean())
```

```python
df[['Age_mean','age']]
```

### Mean impuattion works well when we have normally distributed data

### 2. Median Value Imputation- Used when we have outliers in the dataset

```python
df['Age_median']= df['age'].fillna(df['age'].median())
```

```python
df[['Age_mean','Age_median','age']]
```

### 3. Mode imputation Techniques- for categorical values

```python
df[df['embarked'].isnull()]
```

```python
df['embarked'].unique()
```

```python
mode_val =df[df['embarked'].notna()]['embarked'].mode()[0]
```

```python
df['embarked_mode']=df['embarked'].fillna(mode_val)
```

```python
df[['embarked_mode','embarked']]
```

```python
df['embarked_mode'].isnull().sum()
```
